```python
# Mount Drive BEFORE the runs so each result is saved as soon as it finishes
# (a Colab disconnect then loses at most the run in progress, never a finished one).
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Scaling cell 1/3: n = 20 (equal-learning protocol, complexity-report.md §9.7).
# log.run_name pins the output dir -> runs/ks/ks_n20 (no timestamp to copy by hand).
!python -m jmbc.run exp=ks_n20 log.run_name=ks_n20
```

```python
# Save ks_n20 -> Drive (exact path, idempotent: re-running just re-syncs).
import shutil, os
src = 'runs/ks/ks_n20'
dst = '/content/drive/MyDrive/jax-marl-bc-runs/ks/ks_n20'
assert os.path.exists(src), f"{src} missing - did the ks_n20 run finish?"
shutil.copytree(src, dst, dirs_exist_ok=True)
print(f"saved {src} -> {dst}")
```
